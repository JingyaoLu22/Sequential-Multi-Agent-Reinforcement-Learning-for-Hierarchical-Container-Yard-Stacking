```python
import numpy as np
import gymnasium as gym
from typing import Dict, Tuple, Optional
from enum import Enum
from gymnasium.error import DependencyNotInstalled
from envs.stack_gym import StackEnv

config_dict = {
    "vessel_shape": (5, 4, 4),
    "yard_shape": (5, 4, 4),
    "num_containers": 80,
    "group_num": 4,
    "group_placement": "random",
    "seed": 4307,
}

env = StackEnv(config=config_dict, render_mode="rgb_array")
observation, info = env.reset()
```

```python
print(f"observation.shape: {observation.shape}")
```

```python
import matplotlib.pyplot as plt

# Get the RGB 
img = env.render()

fig_width = env.screen_width / 100  
fig_height = env.screen_height / 100

plt.figure(figsize=(fig_width, fig_height))
plt.imshow(img)
plt.axis('off')  
plt.tight_layout(pad=0)  
plt.show()
```

```python
valid_actions = env._get_valid_yard_actions()
print('Valid actions:', valid_actions)
```

```python
observation, reward, terminated, truncated, info = env.step(96)
print('Reward:', reward)
print('Info:', info)
print('Terminated:', terminated)
print('Current retrieval group:', env.current_retrieval_group)
img = env.render()

plt.figure(figsize=(env.screen_width / 100, env.screen_height / 100))
plt.imshow(img)
plt.axis('off') 
plt.tight_layout(pad=0) 
plt.show()
```
